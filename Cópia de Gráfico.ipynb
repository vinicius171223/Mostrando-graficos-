```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Mês': ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun'],
    'Loja X': [200, 220, 250, 210, 230, 240],
    'Loja Y': [180, 200, 190, 210, 220, 205],
    'Loja Z': [150, 160, 170, 180, 175, 190]
}

lojas_df = pd.DataFrame(data)
lojas_df.set_index('Mês', inplace=True)
print(lojas_df)
```

```python
plt.plot(lojas_df["Loja X"], color = 'blue')
plt.title('Vendas: Loja X',fontsize=18)
plt.xlabel('Meses', fontsize = 13)
plt.ylabel('Vendas em R$', fontsize =13)
plt.style.use('seaborn-v0_8')
plt.show()
```

```python
plt.plot(lojas_df[["Loja X","Loja Y","Loja Z"]])
plt.title('Vendas das Lojas',fontsize=18)
plt.xlabel('Meses', fontsize = 13)
plt.ylabel('Vendas em R$', fontsize =13)
plt.style.use('seaborn-v0_8')
plt.show()
```

```python
fig, ax = plt.subplots()
ax.bar(lojas_df.index, lojas_df["Loja Y"], color = "darkblue")
plt.title('Vendas das Loja Y',fontsize=18)
plt.xlabel('Meses', fontsize = 13)
plt.ylabel('Vendas em R$', fontsize =13)
plt.show()
```

```python
fig, ax = plt.subplots()
ax.bar(lojas_df.index, lojas_df["Loja X"], color ='navy')
ax.bar(lojas_df.index, lojas_df["Loja Y"], color ='darkgreen')
ax.bar(lojas_df.index, lojas_df["Loja Z"], color ='darkred')
plt.title('Vendas das lojas comparadas',fontsize=18)
plt.xlabel('Meses', fontsize = 13)
plt.ylabel('Vendas em R$', fontsize =13)
plt.show()
```

```python
lojas_df[["Loja X","Loja Y","Loja Z"]].sum().plot.pie(autopct='%0.2f%%')
plt.title('Porcentagem de vendas', fontsize = 15)
plt.show()
```

```python
plt.stackplot(lojas_df.index, lojas_df["Loja X"])
plt.stackplot(lojas_df.index, lojas_df["Loja Y"])
plt.stackplot(lojas_df.index, lojas_df["Loja Z"])
plt.show()
```

```python
plt.stackplot(lojas_df.index, lojas_df["Loja X"])
plt.title('Vendas da Loja X',fontsize=18)
plt.xlabel('Meses', fontsize = 13)
plt.ylabel('Vendas em R$', fontsize =13)
plt.show()
```

```python
plt.stackplot(lojas_df.index, lojas_df["Loja Y"], color = 'green')
plt.title('Vendas da Loja Y',fontsize=18)
plt.xlabel('Meses', fontsize = 13)
plt.ylabel('Vendas em R$', fontsize =13)
plt.show()
```

```python
plt.stackplot(lojas_df.index, lojas_df["Loja Z"], color = 'brown')
plt.title('Vendas da Loja Z',fontsize=18)
plt.xlabel('Meses', fontsize = 13)
plt.ylabel('Vendas em R$', fontsize =13)
plt.show()
```

```python
fig, ax = plt.subplots (figsize=(10,6))
ax.scatter(lojas_df["Loja X"], lojas_df["Loja Y"], color = 'green')
plt.title("Encontro de Valores das Lojas X e Y", fontsize = 20)
plt.xlabel("Loja X",fontsize = 17)
plt.ylabel("Loja Y",fontsize = 17)
plt.show()
```

```python
plt.plot(lojas_df["Loja Y"], color = 'green', marker='s', markersize = 12)
plt.style.use('fast')
plt.title("Vendas da Loja Y", fontsize = 20, color = 'red')
plt.xlabel("Meses do ano", fontsize = 17)
plt.ylabel("Vendas em R$", fontsize = 17)
plt.show()
```

```python
fig, ax = plt.subplots()
ax.barh(lojas_df.index, lojas_df["Loja Z"])
plt.title("Gráfico em barra da loja Z", fontsize = 20)
plt.xlabel("Valor em R$", fontsize = 15)
plt.ylabel("Meses do ano", fontsize = 15)
plt.show()
```

```python
fig, ax = plt.subplots()
ax.bar(lojas_df.index, lojas_df["Loja X"])
ax.bar(lojas_df.index, lojas_df["Loja Y"])
plt.title('Vendas das Lojas',fontsize=18)
plt.xlabel('Meses do ano', fontsize = 13)
plt.ylabel('Vendas em R$', fontsize =13)
plt.show()
```
