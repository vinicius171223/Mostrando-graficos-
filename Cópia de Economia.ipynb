#Vendo os 10 Órgãos Superiores com maior Orçamento Realizado

```python
maior_10_sup = df_economia[['NOME ÓRGÃO SUPERIOR', 'ORÇAMENTO REALIZADO (R$)']].groupby('NOME ÓRGÃO SUPERIOR').sum()
maior_10_sup.sort_values(by='ORÇAMENTO REALIZADO (R$)', ascending=False).head(10)
```

#Mostrando a Variação Absoluta e Variação Percentual

```python
vari_absper = df_economia[['NOME ÓRGÃO SUBORDINADO', 'ORÇAMENTO REALIZADO (R$)','ORÇAMENTO INICIAL (R$)']].groupby('NOME ÓRGÃO SUBORDINADO').sum()
vari_absper["Variação Absoluta"] = vari_absper["ORÇAMENTO REALIZADO (R$)"] - vari_absper["ORÇAMENTO INICIAL (R$)"]
vari_absper["Variação Percentual"] = vari_absper["Variação Absoluta"] / vari_absper["ORÇAMENTO INICIAL (R$)"] * 100
vari_absper["Variação Percentual"] = vari_absper["Variação Percentual"].round(2)
display(vari_absper)
```

#Mostrando ranking de eficiência orçamentária

```python
ran_efi = df_economia[['NOME UNIDADE ORÇAMENTÁRIA','% REALIZADO DO ORÇAMENTO']].groupby('NOME UNIDADE ORÇAMENTÁRIA').sum()
ran_efi.sort_values(by='% REALIZADO DO ORÇAMENTO', ascending=False)
```

#Mosrando a despesa média de cada "Nome Função" e quais estão acima ou abaixo da Média

```python
des_med = df_economia[['NOME FUNÇÃO','ORÇAMENTO REALIZADO (R$)']].groupby('NOME FUNÇÃO').mean()
des_med.sort_values(by="ORÇAMENTO REALIZADO (R$)", ascending=False)
soma = df_economia["ORÇAMENTO REALIZADO (R$)"].sum()
media = soma  / 26457
des_med["Acima da média"] = des_med["ORÇAMENTO REALIZADO (R$)"] > media
des_med["Acima da média"] = des_med["Acima da média"].replace(True, "Acima da Média")
des_med["Acima da média"] = des_med["Acima da média"].replace(False, "Abaixo da Média")
des_med
```

#Mostrando qual categoria econômica com mais Orçamento Realizado e a porcentagem

```python
maior_parte = df_economia[["NOME CATEGORIA ECONÔMICA","ORÇAMENTO REALIZADO (R$)"]].groupby("NOME CATEGORIA ECONÔMICA").sum()
total = maior_parte["ORÇAMENTO REALIZADO (R$)"].sum()
maior_parte["Porcentagem"] = maior_parte["ORÇAMENTO REALIZADO (R$)"] / total * 100
maior_parte
```

#Mostrando a proporção entre o realizado e o realizado total em % de cada Órgão Superior

```python
pro_real = df_economia[["NOME ÓRGÃO SUPERIOR","ORÇAMENTO REALIZADO (R$)", "% REALIZADO DO ORÇAMENTO"]].groupby("NOME ÓRGÃO SUPERIOR").sum()
pro_real["Proporção"] = pro_real["ORÇAMENTO REALIZADO (R$)"] / pro_real["% REALIZADO DO ORÇAMENTO"]
pro_real["Proporção"] = pro_real["Proporção"] / 100
pro_real.sort_values(by="Proporção", ascending=False, inplace=True)
pro_real["Proporção"] = pro_real["Proporção"].replace(float('inf'), 0)
pro_real
```

#Vendo qual Órgão Superior teve grande aumento ou não

```python
cor_ou_exp = df_economia[["NOME ÓRGÃO SUPERIOR","ORÇAMENTO INICIAL (R$)","ORÇAMENTO ATUALIZADO (R$)"]].groupby("NOME ÓRGÃO SUPERIOR").sum()
cor_ou_exp["Diferença"] = cor_ou_exp["ORÇAMENTO ATUALIZADO (R$)"] - cor_ou_exp["ORÇAMENTO INICIAL (R$)"]
diferenca_grande = 80000000
cor_ou_exp["Aumento ou não"] = cor_ou_exp["Diferença"] > diferenca_grande
cor_ou_exp["Aumento ou não"] = cor_ou_exp["Aumento ou não"].replace(True, "Grande Aumento")
cor_ou_exp["Aumento ou não"] = cor_ou_exp["Aumento ou não"].replace(False, "Pouco aumento ou diminuiu")
cor_ou_exp
```

#Mostrando qual Unidade Orçamentária utilizou mais ou menos de 50% do Orçamento Atualizado

```python
menos_50_uti = df_economia[["NOME UNIDADE ORÇAMENTÁRIA","ORÇAMENTO ATUALIZADO (R$)","ORÇAMENTO REALIZADO (R$)"]].groupby("NOME UNIDADE ORÇAMENTÁRIA").sum()
Metade = menos_50_uti["ORÇAMENTO ATUALIZADO (R$)"] / 2
menos_50_uti["50% Utilizado"] = Metade >= menos_50_uti["ORÇAMENTO REALIZADO (R$)"]
menos_50_uti["50% Utilizado"] = menos_50_uti["50% Utilizado"].replace(True, "Menos de 50% Utilizado")
menos_50_uti["50% Utilizado"] = menos_50_uti["50% Utilizado"].replace(False, "Mais de ou igual à 50% Utilizado")
menos_50_uti
```
